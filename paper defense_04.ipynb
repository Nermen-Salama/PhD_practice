##the code that use to get the Ottawa dataset

```python

import os
import requests
from google.colab import drive
from urllib.parse import urlparse

# ==========================================
# 1. Mount Google Drive
# ==========================================
drive.mount('/content/drive')

# ==========================================
# 2. Dataset configuration
# ==========================================
DATASET_ID = "y2px5tg92h"
VERSION = 5

BASE_URL = f"https://data.mendeley.com/public-api/datasets/{DATASET_ID}"

OUTPUT_DIR = "/content/drive/MyDrive/Ottawa_MAT_Files"
os.makedirs(OUTPUT_DIR, exist_ok=True)

HEADERS = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/vnd.mendeley-public-dataset.1+json",
    "Referer": f"https://data.mendeley.com/datasets/{DATASET_ID}/{VERSION}"
}

session = requests.Session()
session.headers.update(HEADERS)

# ==========================================
# 3. Retrieve all folders
# ==========================================
print("Retrieving folder structure...")

response = session.get(
    f"{BASE_URL}/folders/{VERSION}",
    timeout=60
)
response.raise_for_status()

folders = response.json()

print("Total folders found:", len(folders))

# ==========================================
# 4. Identify MATLAB folder and descendants
# ==========================================
matlab_folder = next(
    (
        folder for folder in folders
        if "MatLab_Raw_Data_Files" in folder["name"]
    ),
    None
)

if matlab_folder is None:
    raise RuntimeError("MATLAB folder was not found.")

matlab_root_id = matlab_folder["id"]

# Include MATLAB folder and every descendant folder
target_folder_ids = {matlab_root_id}
changed = True

while changed:
    changed = False

    for folder in folders:
        if (
            folder.get("parent_id") in target_folder_ids
            and folder["id"] not in target_folder_ids
        ):
            target_folder_ids.add(folder["id"])
            changed = True

target_folders = [
    folder for folder in folders
    if folder["id"] in target_folder_ids
]

print("\nMATLAB folders identified:")
for folder in target_folders:
    print(" -", folder["name"])

# ==========================================
# 5. Retrieve files from each MATLAB folder
# ==========================================
def get_files(folder_id):
    """Retrieve all files in a folder, including pagination."""

    all_files = []
    offset = 0
    limit = 1000

    while True:
        params = {
            "folder_id": folder_id,
            "version": VERSION,
            "$start": offset,
            "$limit": limit
        }

        response = session.get(
            f"{BASE_URL}/files",
            params=params,
            timeout=60
        )
        response.raise_for_status()

        data = response.json()

        if not isinstance(data, list):
            raise RuntimeError(
                f"Unexpected file response for folder {folder_id}: {data}"
            )

        all_files.extend(data)

        if len(data) < limit:
            break

        offset += limit

    return all_files


all_mat_files = []

for folder in target_folders:
    folder_files = get_files(folder["id"])

    mat_files = [
        f for f in folder_files
        if f.get("filename", "").lower().endswith(".mat")
    ]

    print(
        f"{folder['name']}: "
        f"{len(mat_files)} MAT files"
    )

    for file_info in mat_files:
        file_info["_source_folder"] = folder["name"]

    all_mat_files.extend(mat_files)

print("\n====================================")
print("TOTAL MAT FILES FOUND:", len(all_mat_files))
print("====================================")

if not all_mat_files:
    raise RuntimeError("No MAT files were found.")

# ==========================================
# 6. Download MAT files into their own folders
# ==========================================

downloaded = 0
skipped = 0
failed = []

for index, file_info in enumerate(all_mat_files, start=1):

    filename = os.path.basename(file_info["filename"])
    folder_name = file_info["_source_folder"]

    # Create the corresponding folder in Google Drive
    output_folder = os.path.join(OUTPUT_DIR, folder_name)
    os.makedirs(output_folder, exist_ok=True)

    output_path = os.path.join(output_folder, filename)

    expected_size = file_info.get(
        "content_details", {}
    ).get("size")

    # Skip already completed downloads
    if (
        os.path.exists(output_path)
        and expected_size is not None
        and os.path.getsize(output_path) == expected_size
    ):
        skipped += 1
        print(f"[{index}/{len(all_mat_files)}] Already exists: {folder_name}/{filename}")
        continue

    download_url = file_info.get(
        "content_details", {}
    ).get("download_url")

    if not download_url:
        failed.append((filename, "Missing download URL"))
        continue

    temp_path = output_path + ".part"

    try:
        print(f"\n[{index}/{len(all_mat_files)}] Downloading: {folder_name}/{filename}")

        with session.get(
            download_url,
            stream=True,
            allow_redirects=True,
            timeout=(30, 300)
        ) as response:

            response.raise_for_status()

            with open(temp_path, "wb") as output_file:
                for chunk in response.iter_content(
                    chunk_size=8 * 1024 * 1024
                ):
                    if chunk:
                        output_file.write(chunk)

        actual_size = os.path.getsize(temp_path)

        if expected_size is not None and actual_size != expected_size:
            raise ValueError(
                f"Size mismatch: expected {expected_size}, "
                f"received {actual_size}"
            )

        os.replace(temp_path, output_path)

        downloaded += 1
        print(f"Completed: {folder_name}/{filename}")

    except Exception as error:

        if os.path.exists(temp_path):
            os.remove(temp_path)

        failed.append((filename, str(error)))
        print(f"FAILED: {filename} -- {error}")


# ==========================================
# 7. Final report
# ==========================================

print("\n========== DOWNLOAD SUMMARY ==========")
print("Total MAT files:       ", len(all_mat_files))
print("Downloaded:            ", downloaded)
print("Already existed:       ", skipped)
print("Failed:                ", len(failed))
print("Google Drive location: ", OUTPUT_DIR)

print("\nFiles per folder:")

for folder in target_folders:
    folder_path = os.path.join(OUTPUT_DIR, folder["name"])

    if os.path.isdir(folder_path):
        count = sum(
            1 for name in os.listdir(folder_path)
            if name.lower().endswith(".mat")
        )
        print(f"{folder['name']}: {count} files")

if failed:
    print("\nFailed files:")
    for filename, error in failed:
        print(filename, ":", error)

print("\nFinished.")

```

#Test on Ottawa dataset

These look like real readings rather than errors: they’re whole numbers, and bearings 7 and 8 repeat the same values across files. So the rig probably did run faster on those tests, even though the article says “nominal 1,750 rpm”.

O_7_2 at 2,190 rpm is the one that worries me: the same bearing reads 1,880 in its developing-fault file. If any recorded value is wrong, ECAC would search at the wrong fault frequencies and miss the fault. That would be the data’s fault, not the method’s.

So before deciding, we can check each recorded speed against the vibration signal itself. The shaft-rotation line in the spectrum should sit at the recorded rpm ÷ 60 (about 29 Hz at 1,750 rpm). This cell only reads the files and computes spectra; it doesn’t run the detector:

```python
import numpy as np, scipy.io as sio
from pathlib import Path
ROOT, FS = Path('/content/drive/MyDrive/Ottawa_MAT_Files'), 42000
for f in sorted(ROOT.rglob('*.mat'), key=lambda p: p.stem):
    a = np.asarray(sio.loadmat(f)[f.stem], float)
    x, rpm = a[:, 0] - a[:, 0].mean(), a[0, 2]
    X = np.abs(np.fft.rfft(x * np.hanning(len(x)))); fr = np.fft.rfftfreq(len(x), 1 / FS)
    def peak(lo, hi):
        b = (fr >= lo) & (fr <= hi); i = np.argmax(X[b])
        return fr[b][i], X[b][i] / np.median(X[(fr >= lo - 5) & (fr <= hi + 5)])
    f_rec = rpm / 60
    f1, s1 = peak(0.98 * f_rec, 1.02 * f_rec)        # near the recorded speed (+/-2 %)
    f2, s2 = peak(1.96 * f_rec, 2.04 * f_rec)        # 2x shaft near the recorded speed
    g, sg = peak(20, 45)                             # strongest line 20-45 Hz anywhere
    flag = '' if abs(rpm / 1750 - 1) <= 0.05 else '  <-- >5% from 1750'
    print(f'{f.stem:8s} rec {rpm:6.0f} rpm ({f_rec:5.2f} Hz) | 1x line {f1:5.2f} Hz ({s1:5.1f}x median) '
          f'| 2x {f2:5.2f} Hz ({s2:5.1f}x) | strongest 20-45 Hz {g:5.2f} Hz ({sg:5.1f}x){flag}')
```

##The recorded speed values don’t match the vibration signals. In almost every file, the strongest low-frequency line sits at 29.5–29.8 Hz, about 1,780 rpm, whatever speed is recorded:
Files	Recorded speed	What the spectrum shows
Bearings 7, 8 (H_7, H_8, O_7, O_8)	1,880–2,190 rpm (31–36.5 Hz)	Dominant line at 29.7–29.8 Hz; only noise (≈2–4× median) at the recorded speed
Ball files (B_11–15)	1,709–1,736 rpm (28.5–28.9 Hz)	Dominant line at 29.7 Hz (49–158× median)
Inner race, outer race 6, 9, 10, healthy 1–6	~1,790–1,820 rpm	Line at 29.5–29.8 Hz: 1–2% below the recorded value

This fits the physics of the test rig. The motor is a Lafert LM-71-S4, a 4-pole induction motor. In Canada the mains frequency is 60 Hz, which gives a synchronous speed of 1,800 rpm. A small slip brings it to about 1,780 rpm, which is 29.7 Hz. So the 29.7 Hz line is almost certainly the real shaft speed, and the single recorded value is unreliable: wrong by 1–4% in most files and by 7–25% in bearings 7 and 8.

Why this decides the test. ECAC searches for the fault frequencies within a narrow window around the speed it’s given: ±0.2% with the “measured” setting, ±1% with “nominal”. Feeding it the recorded speeds would make it look in the wrong place in most files. It would fail because of bad data, not because of the method, and the held-out result would be meaningless.

The strongest-line check from my last message would exclude most files under this rule, so it’s not usable. We need a different speed rule, chosen now, before any ECAC run:
